```python
print("======================================")
print("   VALIDADOR DE CONTRASEÑAS IA")
print("======================================")

valida = False

while valida == False:

    contraseña = input("\nIngresa una contraseña: ")

    tiene_mayuscula = False
    tiene_numero = False

    # Verificar longitud
    if len(contraseña) >= 8:

        # Revisar cada caracter
        for caracter in contraseña:

            if caracter.isupper():
                tiene_mayuscula = True

            if caracter.isdigit():
                tiene_numero = True

        # Validar condiciones
        if tiene_mayuscula and tiene_numero:
            valida = True
            print("\n✅ Contraseña válida")
        else:
            print("\n❌ La contraseña debe tener:")
            print("- Al menos una mayúscula")
            print("- Al menos un número")

    else:
        print("\n❌ La contraseña debe tener mínimo 8 caracteres")

print("\nAcceso permitido")
```
